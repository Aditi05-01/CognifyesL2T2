```python
import pandas as pd

# Load dataset
file_path = "Downloads/Dataset .csv"
df = pd.read_csv(file_path)

# Display column names
print("Columns in Dataset:\n", df.columns)
```

```python
# Count occurrences of each price range
most_common_price_range = df["Price range"].value_counts()

print("\nMost Common Price Range:\n", most_common_price_range)

# Visualize
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(8, 5))
sns.barplot(x=most_common_price_range.index, y=most_common_price_range.values, palette="coolwarm")
plt.xlabel("Price Range")
plt.ylabel("Number of Restaurants")
plt.title("Most Common Price Range Among Restaurants")
plt.show()
```

```python
# Convert Price range to string (if necessary)
df["Price range"] = df["Price range"].astype(str)

# Compute average ratings
avg_rating_per_price = df.groupby("Price range")["Aggregate rating"].mean().sort_values(ascending=False)

print("\nAverage Rating for Each Price Range:\n", avg_rating_per_price)

# Visualize
plt.figure(figsize=(8, 5))
sns.barplot(x=avg_rating_per_price.index, y=avg_rating_per_price.values, palette="magma")
plt.xlabel("Price Range")
plt.ylabel("Average Rating")
plt.title("Average Ratings for Different Price Ranges")
plt.show()
```

```python
# Get price range with the highest rating
highest_rated_price_range = avg_rating_per_price.idxmax()
highest_rating = avg_rating_per_price.max()

# Define color palettes
color_palette = sns.color_palette("viridis", len(avg_rating_per_price))

# Map price ranges to colors
price_range_colors = dict(zip(avg_rating_per_price.index, color_palette))

# Get the color associated with the highest rated price range
highest_rating_color = price_range_colors[highest_rated_price_range]

print(f"\nPrice Range with Highest Average Rating: {highest_rated_price_range}")
print(f"Highest Average Rating: {highest_rating}")
print(f"Associated Color: {highest_rating_color}")
```
